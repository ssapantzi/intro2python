## More info on functions: Args and kwargs

You can also use a piece of code that allows you to provide arbitrary arguments (with or without keywords) to your function. If you use `*args` as one of the arguments of your function, this will take any unnamed argument and put all of them in a tuple:

```python
def f(a, *args):
  print(a)
  print(args)
f(1, 2, 3)
```

```python
f(1, 2, 3, 4)
```

```python
# The name args is not important
def f(a, *variable):
  print(a)
  print(variable)
f(1, 2, 3)
f(1, 2, 3, 4)
```

Similarly, if you put two asterisks in front of your variable (e.g. ``**kwargs``), it will assign all names variables to kwargs (in a dictionary format).

```python
def f(a, **kwargs):
  print(a)
  print(kwargs)
f(a=1, b=2, c=3)
```

```python
#4
def has22 (list_c: list):
    """The input of this function is a list of integers. The function checks if there are two '2's next to each other on the list.
    If there are, it will return True. If not, it will return False."""
    true_events = []
    for n in range (len(list_c)- 1):
        if list_c[n] == 2 and list_c[n + 1] == 2:
            true_events.append(1)
    if len(true_events)> 0:
        print ('True')
    else: 
        print ('False')


has22([1,3,2,3,1,2])
```

## Problem 5

You and your friends are really into the newest season of some reality TV show where everyone speed-dates each other, and you are trying to think of couple names (ship names) for several pairs. A common heuristic for coming up with couple names is combining the first syllable of one person's name with the LAST syllable of the other person's name. So, for example, the couple name for `Sharon` and `Janet` would be `Shanet`, combining the first syllable `Sha` from Sharon and the last syllable `net` from `Janet`.
 

Rules for syllable parsing:
If a vowel (letters aeiou and **sometimes** y) is followed by a consonant, that vowel marks the end of the syllable. (e.g. in `Sharon`, `a` is followed by `r`, and thus, the first syllable is `Sha`). EXCEPT, if `sh` follows a vowel, then `sh` marks the end of that syllable. e.g. the first syllable of `Ashley` is `Ash`, not `As`
 

However, what if you have a couple with the names `Aaron` and `Sharon`? Combining the first syllable and last syllable from those two names would yield `Aaron`, and switching the order (where we take the first syllable from Sharon, and the last syllable from Aaron) would yield `Sharon`. These are BORING couple names, because the output is the same as one of the names of the people in the couple. In these cases, you would want to combine the first syllable from `Aaron` with the first syllable from `Sharon`, thus yielding `AaSha`, a much better couple name. [If it makes it easier, I will accept `Ronron` as an answer too.]

```python
def name_generator(person1 : str, person2 : str) -> str:
    ##### Put your work below this line.
    # first we need to slice the names correclty
    #then check if the generated name is different than the original names
    #if not, combine a different part of the name together
    #if vowel is followed by consonant, it is the end of the syllable
    vowels = ['a','e','i','o','u', 'A', 'E', 'I', 'O', 'U', 'Y', 'y']
    short_name_1 = [] #lists of the letters that go into the first syllable 
    short_name_2 = [] #lists of the letters in the las syllable
    boring_name_2 = [] # in case first syllable of person 1 == last syllable of person 2, use the first syllable of person 2
    #first name, reading forward
    for n in range(len(person1) -1):
        short_name_1.append(person1[n])
        if person1[n] in vowels and person1[n+1] == 's' and person1[n+2] == 'h':
            short_name_1.append(person1[n+1])
            short_name_1.append(person1[n+2])
            break
        elif person1[n] in vowels and person1 [n + 1] not in vowels:
            break
        else:
            continue
    short_1 = ''.join(short_name_1) #turn list into string
    #print(short_1)

#second name, reading backward
    rev_person2 = person2[:: -1]
    #print(rev_person2)
    for i in range(1, len(rev_person2)-1): 
        short_name_2.append(rev_person2[i])
        if rev_person2[i] not in vowels and rev_person2 [i+1] in vowels:
            break
        elif rev_person2[i] not in vowels and rev_person2 [i+1] not in vowels:
            short_name_2.append(rev_person2[i+1])
            break
        else:
            continue

    #make list into a sting, reverse the string, add back the first letter
    short_2_rev = ''.join(short_name_2) #turn list into string
    short_2 = short_2_rev [:: -1] + rev_person2[0]
    #print (short_2)
    
    #alternative for boring name, where the same rules apply in person 2 to pick first part of the name
    for n in range(len(person2) -1):
        boring_name_2.append(person2[n])
        if person2[n] in vowels and person2[n+1] == 's' and person2[n+2] == 'h':
            boring_name_2.append(person2[n+1])
            boring_name_2.append(person2[n+2])
            break
        elif person2[n] in vowels and person2 [n + 1] not in vowels:
            break
        else:
            continue
    boring_2 = ''.join(boring_name_2) #turn list into string
    #print(boring_2)
    couple_name = (short_1 + short_2) #merge strings together
    if couple_name == person1 or couple_name == person2:
        couple_name = (short_1 + boring_2.lower())
    return (couple_name)
    #### Put your work above this line 
    

name_generator('Yonathan', 'Betsy')
```
